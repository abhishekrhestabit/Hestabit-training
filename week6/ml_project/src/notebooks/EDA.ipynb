```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the PROCESSED data (or RAW if you want to compare)
bf = pd.read_csv('../data/raw/dataset.csv')
df = pd.read_csv('../data/processed/final.csv')
print(df.info())
print(df.head())
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(bf.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap (Should be empty)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap (Should be empty)")
plt.show()
```

```python
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=['float64', 'int64'])
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Feature Correlation Matrix")
plt.show()
```

```python
# CHANGE 'TargetColumn' to your label column name
target_col = 'Survived' 

if target_col in df.columns:
    sns.countplot(x=target_col, data=df)
    plt.title("Target Label Distribution")
    plt.show()
else:
    print(f"Column {target_col} not found. Please update the variable.")
```
